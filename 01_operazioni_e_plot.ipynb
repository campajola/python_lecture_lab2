### Funzioni speciali per creare un numpy array

`NumPy` include un'ampia serie di funzioni per la creazione di array. Ora vedremo alcune delle più comunemente utilizzate.

La prima è `np.arange` (che significa "array range"), la quale funziona in modo molto simile alla funzione standard `range()` di Python, inclusa la sua modalità predefinita di partire da zero, escludere il numero più alto nell'intervallo e la possibilità di specificare un "passo".

````python
np.arange(2, 8, 2)
````

````python
np.arange(5, 20, 2) #0 .. n-1  (!)
````

````python
np.arange(1, 9, 2) # start, end (exclusive), step
````

In aggiunta, abbiamo `np.linspace` (che significa "spazio lineare"), che genera una serie di numeri a partire dal primo argomento fino al secondo argomento. Il terzo argomento definisce quanti numeri creare.

````python
np.linspace(0, 1, 6)   # start, end, num-points
````

Nota come includa il punto finale, a differenza di `arange()`. Puoi modificare questa caratteristica utilizzando l'argomento `endpoint`.

````python
np.linspace(0, 1, 5, endpoint=False)
````

````python
plt.errorbar(I_m, V_m, xerr=err_I, yerr=err_V, fmt='*', markersize=2., capsize=1.5, label='dati (conf. a monte) *')

plt.errorbar(I_v, V_v, xerr=err_I, yerr=err_V, fmt='s', markersize=2., capsize=1.5, label='dati (conf. a valle) *')

plt.annotate('*err. massimi',
        xy=(0.52, 0.82), xycoords='axes fraction',
        ha='left', va='top', fontsize=10)


plt.legend()
plt.ylabel('$\Delta$V [V]', size=14)
plt.xlabel('I [$\mu$A]', size=14)
plt.grid()
plt.title('Stima di R con il metodo voltamperometrico')
plt.xlim(0, 400)
plt.ylim(0, 1.2)
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Definizione della funzione di risposta dell'impulso con due decadimenti
def scintillator_signal_dual_decay(t, A1, A2, t0, tau_r, tau_d1, tau_d2):

    heaviside = np.heaviside(t - t0, 1)
    # Modello del segnale con due tempi di decadimento
    signal = (A1 * (np.exp(-(t - t0) / tau_d1) - np.exp(-(t - t0) / tau_r)) + 
              A2 * (np.exp(-(t - t0) / tau_d2) - np.exp(-(t - t0) / tau_r))) * heaviside
    return signal

A1 = 5.0      # ampiezza della componente rapida
A2 = 1      # ampiezza della componente lenta
t0 = 10       # Tempo di arrivo del segnale (in nanosecondi)
tau_r = 1   # Tempo di salita (in nanosecondi)
tau_d1 = 10.0  # Primo tempo di discesa (in nanosecondi, rapido)
tau_d2 = 50.0 # Secondo tempo di discesa (in nanosecondi, lento)

t = np.linspace(0, 200, 1000)  # Tempo da 0 a 50 ns
signal = scintillator_signal_dual_decay(t, A1, A2, t0, tau_r, tau_d1, tau_d2)

plt.plot(t, signal)
plt.title("Segnale di uno scintillatore (due tempi di decadimento)")
plt.xlabel("Tempo [ns]")
plt.ylabel("Ampiezza del segnale [a.u.]")
plt.grid(True)
#plt.yscale('log')
plt.show()
````

### Istogrammi

Supponiamo di avere un insieme di dati che rappresenta le misure di energia raccolte da un rilevatore. Vogliamo visualizzare la distribuzione di queste misure in un istogramma.

La funzione `plt.hist()` serve per creare un istogramma, cioè un grafico che mostra come sono distribuiti i valori in un insieme di dati.

````python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 10000) # Generiamo un esempio di dati (es. misure di energia da un rilevatore)
data
````
